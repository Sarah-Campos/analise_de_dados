### Passo 3 - Criar o notebook

### Passo 4 - Ler a tabela

Primeiro importo o pandas, que é a biblioteca que ajuda a trabalhar com tabelas.

```python
# importando a biblioteca pandas
import pandas as pd

# lendo o arquivo csv que está na mesma pasta
tabela = pd.read_csv("tabela_populacao.csv")

# mostrando as primeiras linhas pra ver se deu certo
tabela.head()
```

```python
# olhando quantas linhas e colunas a tabela tem
print("Número de linhas:", len(tabela))
print("Colunas:", tabela.columns)
```

### Passo 5 - Somar a população por estado

Cada estado (UF) tem vários municípios. Preciso somar a população de todos os
municípios de cada estado, tanto de 2010 quanto de 2022.

```python
# o groupby junta as linhas que têm o mesmo valor na coluna UF
# e o sum() soma os valores de população dentro de cada grupo
tabela_estado = tabela.groupby("UF")[["POP_2010", "POP_2022"]].sum()

# o groupby deixa a coluna UF como índice, então trago ela de volta como coluna normal
tabela_estado = tabela_estado.reset_index()

tabela_estado
```

### Passo 6 - Descobrir quem mais cresceu (2010 para 2022)

Para saber o crescimento, é só pegar a população de 2022 e subtrair a de 2010.
Depois ordeno a tabela do estado que mais cresceu para o que menos cresceu.

```python
# criando uma coluna nova com a diferença (crescimento)
tabela_estado["crescimento"] = tabela_estado["POP_2022"] - tabela_estado["POP_2010"]

# ordenando pela coluna crescimento, do maior para o menor
tabela_estado = tabela_estado.sort_values("crescimento", ascending=False)

tabela_estado
```

### Passo 7 - Salvar essa tabela em um arquivo CSV novo

```python
# salvando a tabela por estado num arquivo novo, na mesma pasta do projeto
# index=False é pra não salvar aquela coluna de numeração que o pandas cria sozinho
tabela_estado.to_csv("populacao_por_estado.csv", index=False)

print("Tabela por estado salva!")
```

### Passo 8 - Agora fazendo por município

Dessa vez não preciso somar nada, porque cada linha da tabela original já é
um município. Só vou pegar as colunas que interessam.

```python
# escolhendo as colunas que eu quero usar
tabela_municipio = tabela[["UF", "NOME_MUNICIPIO", "POP_2010", "POP_2022"]].copy()

tabela_municipio.head()
```

### Passo 9 - Calcular o crescimento de cada município e ordenar

```python
# mesma ideia de antes: 2022 menos 2010
tabela_municipio["crescimento"] = tabela_municipio["POP_2022"] - tabela_municipio["POP_2010"]

# ordenando do município que mais cresceu para o que menos cresceu
tabela_municipio = tabela_municipio.sort_values("crescimento", ascending=False)

tabela_municipio
```

### Passo 10 - Salvar a tabela por município em CSV

```python
tabela_municipio.to_csv("populacao_por_municipio.csv", index=False)

print("Tabela por município salva!")
print("Pronto, terminei!")
```
